```python
import json
import re
from typing import Optional
import random

SYSTEM_PROMPT = (
    "You are a research question decomposition agent. Given a research "
    "question, break it into topic-grouped sub-questions with detail "
    "questions, and provide search ready queries."
)
# Matches a leading "- " or " - " bullet artifact left over from the source data
DASH_PREFIX_RE = re.compile(r"^\s*-\s*")
# Matches leftover unrendered template tags like <query1>...</query1>
TEMPLATE_TAG_RE = re.compile(r"</?query\d*>", re.IGNORECASE)
RANDOM_SEED = 42
TRAIN_FRACTION = 0.90
VAL_FRACTION = 0.05
# remaining ~0.05 goes to test
```

```python
def clean_text(text: str) -> str:
    """Strip leading bullet artifacts and stray whitespace from one string."""
    text = DASH_PREFIX_RE.sub("", text)
    return text.strip()


def has_template_artifact(items) -> bool:
    """Recursively check a (possibly nested) list of strings for leftover
    <queryN> style template tags, which indicate a malformed source row."""
    for item in items:
        if isinstance(item, str) and TEMPLATE_TAG_RE.search(item):
            return True
        if isinstance(item, list) and has_template_artifact(item):
            return True
    return False
```

```python
def build_sub_questions(headers) -> list:
    """Convert the nested `headers` column (list of [topic, *details])
    into [{"topic": ..., "details": [...]}, ...], with bullets cleaned."""
    sub_questions = []
    for group in headers:
        if not group:
            continue
        cleaned = [clean_text(item) for item in group if clean_text(item)]
        if not cleaned:
            continue
        topic, *details = cleaned
        sub_questions.append({"topic": topic, "details": details})
    return sub_questions


def row_to_example(row: dict) -> Optional[dict]:
    """Convert one raw dataset row into one chat-format training example.
    Returns None if the row fails quality checks and should be dropped."""
    question = (row.get("question") or "").strip()
    headers = row.get("headers") or []
    queries = row.get("queries") or []

    # Drop rows missing any of the three fields we need
    if not question or not headers or not queries:
        return None

    # Drop rows containing leftover template artifacts
    if has_template_artifact(headers) or has_template_artifact(queries):
        return None

    sub_questions = build_sub_questions(headers)
    if not sub_questions:
        return None

    cleaned_queries = [clean_text(q) for q in queries if clean_text(q)]
    if not cleaned_queries:
        return None

    assistant_payload = {
        "sub_questions": sub_questions,
        "search_queries": cleaned_queries,
    }

    return {
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": question},
            {
                "role": "assistant",
                "content": json.dumps(assistant_payload, ensure_ascii=False),
            },
        ]
    }
```

```python
def generate_prediction(question: str, max_new_tokens: int = 768):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": question},
    ]

    rendered_text = tokenizers.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    rendered_text += "<think>\n\n</think>\n\n"

    inputs = tokenizers(text=rendered_text, return_tensors="pt").to("cuda")

    outputs = model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=max_new_tokens,
        temperature=0.3,
        top_p=0.9,
        do_sample=True,
    )

    raw_output = tokenizers.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    return raw_output
```

```python
import json
import difflib
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from tqdm import tqdm
import os


embedder = SentenceTransformer('BAAI/bge-base-en-v1.5')


def extract_json_from_output(raw_output: str):
    raw_output = raw_output.strip()
    decoder = json.JSONDecoder()
    start = raw_output.find("{")
    if start == -1:
        return None
    try:
        obj, _ = decoder.raw_decode(raw_output[start:])
        return obj
    except json.JSONDecodeError:
        return None


def flatten_topics_only(parsed_json):
    if not parsed_json or "sub_questions" not in parsed_json:
        return []
    topics = []
    for item in parsed_json["sub_questions"]:
        if isinstance(item, dict) and "topic" in item:
            topics.append(item["topic"])
        elif isinstance(item, str):
            topics.append(item)
    return topics


def flatten_details_only(parsed_json):
    if not parsed_json or "sub_questions" not in parsed_json:
        return []
    all_details = []
    for item in parsed_json["sub_questions"]:
        if isinstance(item, dict):
            details = item.get("details", [])
            if isinstance(details, list):
                all_details.extend(str(d) for d in details)
    return all_details


def flatten_queries(parsed_json):
    if not parsed_json or "search_queries" not in parsed_json:
        return []
    return [q for q in parsed_json["search_queries"] if isinstance(q, str)]


def embedding_cosine_similarity(text_a: str, text_b: str) -> float:
    if not text_a.strip() or not text_b.strip():
        return 0.0
    embeddings = embedder.encode([text_a, text_b], normalize_embeddings=True)
    return float(cosine_similarity([embeddings[0]], [embeddings[1]])[0][0])


def text_overlap_ratio(text_a: str, text_b: str) -> float:
    return difflib.SequenceMatcher(None, text_a.lower(), text_b.lower()).ratio()


def evaluate_single(question: str, generated_raw: str, reference_raw: str):
    gen_parsed = extract_json_from_output(generated_raw)
    ref_parsed = extract_json_from_output(reference_raw)

    result = {
        "question": question,
        "valid_json": gen_parsed is not None,
    }

    if gen_parsed is None or ref_parsed is None:
        result.update({
            "topic_similarity": 0.0,
            "details_similarity": 0.0,
            "query_similarity": 0.0,
            "topic_count_match": False,
            "query_count_match": False,
        })
        return result

    gen_topics = flatten_topics_only(gen_parsed)
    ref_topics = flatten_topics_only(ref_parsed)
    gen_details = flatten_details_only(gen_parsed)
    ref_details = flatten_details_only(ref_parsed)
    gen_queries = flatten_queries(gen_parsed)
    ref_queries = flatten_queries(ref_parsed)

    gen_topics_text = " ".join(gen_topics)
    ref_topics_text = " ".join(ref_topics)
    gen_details_text = " ".join(gen_details)
    ref_details_text = " ".join(ref_details)
    gen_queries_text = " ".join(gen_queries)
    ref_queries_text = " ".join(ref_queries)

    result.update({
        "num_sub_questions_generated": len(gen_topics),
        "num_sub_questions_reference": len(ref_topics),
        "num_details_generated": len(gen_details),
        "num_details_reference": len(ref_details),
        "num_search_queries_generated": len(gen_queries),
        "num_search_queries_reference": len(ref_queries),
        "topic_count_match": len(gen_topics) == len(ref_topics),
        "query_count_match": len(gen_queries) == len(ref_queries),
        "topic_similarity": embedding_cosine_similarity(gen_topics_text, ref_topics_text),
        "details_similarity": embedding_cosine_similarity(gen_details_text, ref_details_text),
        "query_similarity": embedding_cosine_similarity(gen_queries_text, ref_queries_text),
        "topic_text_overlap": text_overlap_ratio(gen_topics_text, ref_topics_text),
        "details_text_overlap": text_overlap_ratio(gen_details_text, ref_details_text),
        "query_text_overlap": text_overlap_ratio(gen_queries_text, ref_queries_text),
        "generated_topics": gen_topics,
        "reference_topics": ref_topics,
        "generated_details": gen_details,
        "reference_details": ref_details,
        "generated_queries": gen_queries,
        "reference_queries": ref_queries,
    })
    return result


def run_evaluation(test_dataset, num_samples=None, verbose=True,
                    checkpoint_path="eval_checkpoint.jsonl", resume=True):
    """
    Saves each result incrementally to checkpoint_path (JSONL, one result per line).
    If resume=True and checkpoint_path exists, skips already-evaluated questions
    and continues from where it left off.
    """
    n = len(test_dataset) if num_samples is None else min(num_samples, len(test_dataset))

    # --- Load already-completed results if resuming ---
    completed_questions = set()
    results = []
    if resume and os.path.exists(checkpoint_path):
        with open(checkpoint_path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                try:
                    result = json.loads(line)
                    results.append(result)
                    completed_questions.add(result["question"])
                except json.JSONDecodeError:
                    continue  # skip corrupted last line (e.g. from a crash mid-write)
        if verbose and results:
            print(f"Resuming: found {len(results)} already-evaluated examples in {checkpoint_path}")

    # Open in append mode so each new result is written immediately
    with open(checkpoint_path, "a", encoding="utf-8") as f:
        for i in tqdm(range(n), desc="Evaluating"):
            row = test_dataset[i]
            question = row["messages"][1]["content"]
            reference_raw = row["messages"][2]["content"]

            if question in completed_questions:
                continue  # already done, skip re-running

            generated_raw = generate_prediction(question)
            eval_result = evaluate_single(question, generated_raw, reference_raw)
            eval_result["generated_raw"] = generated_raw
            eval_result["reference_raw"] = reference_raw

            # Write immediately and flush to disk — survives a crash right after this line
            f.write(json.dumps(eval_result, ensure_ascii=False) + "\n")
            f.flush()
            os.fsync(f.fileno())

            results.append(eval_result)
            completed_questions.add(question)

            if verbose:
                print(f"\n[{i}] {question[:60]}...")
                print(f"  Valid JSON: {eval_result['valid_json']}")
                if eval_result['valid_json']:
                    print(f"  Topic similarity:   {eval_result['topic_similarity']:.3f}")
                    print(f"  Details similarity: {eval_result['details_similarity']:.3f}")
                    print(f"  Query similarity:   {eval_result['query_similarity']:.3f}")
                    print(f"  Topic count match:  {eval_result['topic_count_match']} "
                      f"({eval_result['num_sub_questions_generated']} vs {eval_result['num_sub_questions_reference']})")
                    print(f"  Query count match:  {eval_result['query_count_match']} "
                      f"({eval_result['num_search_queries_generated']} vs {eval_result['num_search_queries_reference']})")

    # --- Summary over everything completed so far (resumed + new) ---
    valid = [r for r in results if r["valid_json"]]
    n_valid = len(valid)
    n_total = len(results)

    print("\n" + "=" * 60)
    print(f"Total evaluated so far:      {n_total}")
    print(f"Valid JSON rate:             {n_valid}/{n_total} ({100*n_valid/n_total:.1f}%)" if n_total else "No results yet")
    if n_valid > 0:
        avg_topic_sim = sum(r["topic_similarity"] for r in valid) / n_valid
        avg_details_sim = sum(r["details_similarity"] for r in valid) / n_valid
        avg_query_sim = sum(r["query_similarity"] for r in valid) / n_valid
        topic_count_match_rate = sum(r["topic_count_match"] for r in valid) / n_valid
        query_count_match_rate = sum(r["query_count_match"] for r in valid) / n_valid
        print(f"Avg topic similarity:        {avg_topic_sim:.3f}")
        print(f"Avg details similarity:      {avg_details_sim:.3f}")
        print(f"Avg query similarity:        {avg_query_sim:.3f}")
        print(f"Topic count match rate:      {100*topic_count_match_rate:.1f}%")
        print(f"Query count match rate:      {100*query_count_match_rate:.1f}%")
    print("=" * 60)

    return results
```

```python
eval_results = run_evaluation(test_dataset, verbose=True)

with open("eval_results.json", "w", encoding="utf-8") as f:
    json.dump(eval_results, f, indent=2, ensure_ascii=False)

print("\nSaved to eval_results.json")
```

```python
def extract_json_from_output(raw_output: str):
    """Handle a leading reasoning block ending in </think> (opening tag may be
    absent from generated text), then extract just the first valid JSON object,
    ignoring any hallucinated text that follows it."""
    
    # Step 1: cut everything up to and including a closing </think>, if present
    if "</think>" in raw_output:
        raw_output = raw_output.split("</think>", 1)[1]
    raw_output = raw_output.strip()

    # Step 2: extract only the first valid JSON object, ignoring trailing text
    decoder = json.JSONDecoder()
    # Find the first '{' to start decoding from
    start = raw_output.find("{")
    if start == -1:
        return None, raw_output  # no JSON found at all

    try:
        obj, end_index = decoder.raw_decode(raw_output[start:])
        return obj, raw_output
    except json.JSONDecodeError:
        return None, raw_output


def generate_sub_questions(question: str, max_new_tokens: int = 512):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": question},
    ]

    rendered_text = tokenizers.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    # Fixed: explicit text= keyword, since this is a multimodal processor
    inputs = tokenizers(text=rendered_text, return_tensors="pt").to("cuda")

    outputs = model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=max_new_tokens,
        temperature=0.3,
        top_p=0.9,
        do_sample=True,
    )

    raw_output = tokenizers.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    parsed, cleaned_text = extract_json_from_output(raw_output)

    if parsed is not None:
        return parsed
    else:
        print("Warning: could not parse JSON. Raw model output was:")
        print(raw_output)
        return {"raw_output": raw_output, "sub_questions": None, "search_queries": None}

# Example usage
question = "What is quantum computing and how does it differ from classical computing? how we can apply quantum computing in machine learning and artificial intelligence? what are the challenges and limitations of quantum computing in practical applications?"
result = generate_sub_questions(question)
print(json.dumps(result, indent=2, ensure_ascii=False))
```

```python
import torch
import gc

# Delete all commonly-named variables that might hold GPU tensors
for var_name in [
    "model", "tokenizer", "tokenizers", "trainer", "trainer_stats",
    "inputs", "outputs", "raw_output", "smoke_trainer",
    "optimizer", "scheduler", "accelerator", "result", "parsed"
]:
    if var_name in globals():
        del globals()[var_name]

# Force multiple garbage collection passes
for _ in range(3):
    gc.collect()

# Clear CUDA cache and synchronize
torch.cuda.empty_cache()
torch.cuda.ipc_collect()
torch.cuda.synchronize()

# Check actual GPU memory via nvidia-smi (more reliable than torch's own counters)
print(f"Torch allocated: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
print(f"Torch reserved:  {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
```
